```python
import pandas as pd 
```

```python
data = pd.read_csv("/Users/thas/Desktop/Churn_Modelling.csv")
```

```python
from sklearn.preprocessing import LabelEncoder

geo_encoder = LabelEncoder()
data['Geography'] = geo_encoder.fit_transform(data['Geography'])

gender_encoder = LabelEncoder()
data['Gender'] = gender_encoder.fit_transform(data['Gender'])
```

```python
data['BalanceSalaryRatio'] = data['Balance'] / (data['EstimatedSalary'] + 1)

data['TenurePerProduct'] = data['Tenure'] / (data['NumOfProducts'] + 1)

data['AgeActiveInteraction'] = data['Age'] * data['IsActiveMember']
```

```python
print(data[['BalanceSalaryRatio', 'TenurePerProduct', 'AgeActiveInteraction', 'Exited']].corr())
```

```python
data.to_csv("/Users/thas/Desktop/Churn_Modelling.csv", index=False)
```
